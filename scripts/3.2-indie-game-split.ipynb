```python
import pandas as pd
import ast
```

```python
FINAL_CSV_PATH = "../data/gold/unified.csv"

final_df = pd.read_csv(FINAL_CSV_PATH)
```

```python
print(dict(final_df.dtypes.apply(lambda x: str(x))))
```

```python
def contains_indie(cat_list):
    if isinstance(cat_list, list):
        return any("indi" in str(x).lower() for x in cat_list)
    return False

def parse_list(x):
    if pd.isna(x):
        return []
    try:
        return ast.literal_eval(x)
    except:
        return []
```

```python
for col in ["categories_steam", "categories_epic", "categories_igdb"]:
    final_df[col] = final_df[col].apply(parse_list)

final_df["is_indie"] = (
    final_df["categories_steam"].apply(contains_indie)
    | final_df["categories_epic"].apply(contains_indie)
    | final_df["categories_igdb"].apply(contains_indie)
)

indie_games = final_df[final_df["is_indie"]]
```

```python
indie_games.head()
```

```python
indie_games.describe()
```

```python
indie_games = indie_games.drop(columns=['Unnamed: 0'], errors='ignore')
indie_games = indie_games.drop(columns=['is_indie'], errors='ignore')
indie_games.to_csv('../data/gold/indie.csv')
```

```python
notindie_games = final_df[~final_df["is_indie"]]
notindie_games.head()
```

```python
notindie_games.describe()
```

```python
notindie_games = notindie_games.drop(columns=['Unnamed: 0'], errors='ignore')
notindie_games = notindie_games.drop(columns=['is_indie'], errors='ignore')
notindie_games.to_csv('../data/gold/notindie.csv')
```

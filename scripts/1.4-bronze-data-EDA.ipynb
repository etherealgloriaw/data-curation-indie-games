```python
import json
from pathlib import Path
import pandas as pd
```

```python
def infer_type(value):
    if isinstance(value, dict):
        return {k: infer_type(v) for k, v in value.items()}
    elif isinstance(value, list):
        if len(value) == 0:
            return "list"
        return [infer_type(value[0])]   # assume all items same shape
    else:
        return type(value).__name__

def get_csv_schema_with_cast(path: str):
    df = pd.read_csv(path)

    # Cast object → string
    object_cols = df.select_dtypes(include="object").columns
    df = df.astype({col: "string" for col in object_cols})

    # Cast date-like columns → datetime64
    for col in df.columns:
        if "date" in col.lower():
            df[col] = pd.to_datetime(df[col], errors="ignore")

    schema = df.dtypes.apply(lambda x: x.name).to_dict()
    return schema
```

```python
#EPIC
path = Path(r"../data/bronze/epic/epic_games_all.json")
text = path.read_text(encoding="utf-8").strip()
data = json.loads(text)
schema = infer_type(data)
print(schema)
```
